# ERA5 Temperature Pipeline
**Stages:** CDS API → GRIB download → pandas DataFrame → PostgreSQL (Docker)

### Prerequisites
```bash
pip install cdsapi cfgrib xarray pandas sqlalchemy psycopg2-binary
```
And run your Postgres container:
```bash
docker run -d \
  --name era5_pg \
  -e POSTGRES_USER=era5 \
  -e POSTGRES_PASSWORD=era5pass \
  -e POSTGRES_DB=climate \
  -p 5432:5432 \
  postgres:15
```

## Stage 1 — Download from CDS API

````python
import cdsapi

client = cdsapi.Client()

dataset = 'reanalysis-era5-pressure-levels'
request = {
    'product_type': ['reanalysis'],
    'variable': ['temperature'],
    'year': ['2026'],
    'month': ['04'],
    'day': ['01'],
    'time': ['13:00'],
    'pressure_level': ['1000'],
    'data_format': 'grib',
}
target = 'download.grib'

client.retrieve(dataset, request, target)
print(f'Downloaded: {target}')
````

## Stage 2 — GRIB → xarray Dataset → pandas DataFrame

````python
import cfgrib
import xarray as xr
import pandas as pd

# Open GRIB file as an xarray Dataset
# filter_by_keys selects the temperature field (shortName 't')
ds = xr.open_dataset(
    target,
    engine='cfgrib',
    backend_kwargs={'filter_by_keys': {'shortName': 't'}}
)

print(ds)
````

````python
# Convert to a flat DataFrame
df = ds.to_dataframe().reset_index()

# Rename columns to something readable
df = df.rename(columns={
    'latitude':      'lat',
    'longitude':     'lon',
    'valid_time':    'valid_time',
    'isobaricInhPa': 'pressure_hpa',
    't':             'temperature_k',   # Kelvin
})

# Add a convenience column in Celsius
df['temperature_c'] = df['temperature_k'] - 273.15

# Drop rows with no temperature value (edges of grid)
df = df.dropna(subset=['temperature_k'])

print(f'Shape: {df.shape}')
df.head()
````

````python
# Quick sanity check
df.describe()
````

## Stage 3 — Stream DataFrame to PostgreSQL

We use **SQLAlchemy** + `to_sql` with `chunksize` so large datasets are streamed in batches rather than loaded into memory all at once.

````python
from sqlalchemy import create_engine, text

# ── connection config ──────────────────────────────────────────
PG_USER     = 'era5'
PG_PASSWORD = 'era5pass'
PG_HOST     = 'localhost'
PG_PORT     = 5432
PG_DB       = 'climate'
TABLE_NAME  = 'era5_temperature'
# ──────────────────────────────────────────────────────────────

conn_str = f'postgresql+psycopg2://{PG_USER}:{PG_PASSWORD}@{PG_HOST}:{PG_PORT}/{PG_DB}'
engine   = create_engine(conn_str, pool_pre_ping=True)

# Quick connection test
with engine.connect() as conn:
    result = conn.execute(text('SELECT version()'))
    print(result.fetchone()[0])
````

````python
import math

CHUNK_SIZE = 10_000   # rows per INSERT batch

# Columns we want to persist
cols_to_write = ['valid_time', 'pressure_hpa', 'lat', 'lon',
                 'temperature_k', 'temperature_c']

df_out = df[cols_to_write].copy()

total_chunks = math.ceil(len(df_out) / CHUNK_SIZE)
print(f'Streaming {len(df_out):,} rows in {total_chunks} chunks of {CHUNK_SIZE:,}…')

df_out.to_sql(
    name      = TABLE_NAME,
    con       = engine,
    if_exists = 'replace',   # use 'append' on subsequent runs
    index     = False,
    chunksize = CHUNK_SIZE,
    method    = 'multi',     # batched INSERT for speed
)

print('Done ✓')
````

## Stage 4 — Verify & Explore from Postgres

````python
# Read back a sample to confirm the write
df_check = pd.read_sql(
    f"SELECT * FROM {TABLE_NAME} LIMIT 5",
    con=engine
)
df_check
````

````python
# Row count in Postgres
with engine.connect() as conn:
    n = conn.execute(text(f'SELECT COUNT(*) FROM {TABLE_NAME}')).fetchone()[0]
print(f'Rows in Postgres: {n:,}')
````

````python
# Example analytical query — mean temperature per pressure level
df_stats = pd.read_sql(f"""
    SELECT
        pressure_hpa,
        ROUND(AVG(temperature_c)::numeric, 2) AS avg_temp_c,
        ROUND(MIN(temperature_c)::numeric, 2) AS min_temp_c,
        ROUND(MAX(temperature_c)::numeric, 2) AS max_temp_c
    FROM {TABLE_NAME}
    GROUP BY pressure_hpa
    ORDER BY pressure_hpa
""", con=engine)

df_stats
````
